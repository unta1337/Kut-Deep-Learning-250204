# b_tensor_initialization_copy

```python
l1 = [1, 2, 3]
t1 = torch.Tensor(l1)

l2 = [1, 2, 3]
t2 = torch.tensor(l2)

l3 = [1, 2, 3]
t3 = torch.as_tensor(l3)

# 각각의 텐서는 원본 배열을 복사해 생성됨.
# 즉, 원본을 수정해도 복사된 텐서는 변하지 않음.
l1[0] = 100
l2[0] = 100
l3[0] = 100

print(t1)
print(t2)
print(t3)
```

```python
l4 = np.array([1, 2, 3])
t4 = torch.Tensor(l4)

l5 = np.array([1, 2, 3])
t5 = torch.tensor(l5)

l6 = np.array([1, 2, 3])
t6 = torch.as_tensor(l6)

# 넘파이 배열도 기본적으로 복사 후 텐서를 생성함.
# 단, as_tensor로 불러오면 참조로 생성함.
# 즉, 원본 넘파이 배열을 수정하면 텐서도 변경됨.
l4[0] = 100
l5[0] = 100
l6[0] = 100

print(t4)
print(t5)
print(t6)
```

# e_tensor_type_conversion

```python
a = torch.ones((2, 3))                                  # 기본 자료형: float32
                                                        # 기본 자료형은 전역 기본 자료형을 사용한다.
                                                        # 이때 전역 기본 자료형은 최초 설치 시 float32로 설정된다.
                                                        # ref: https://docs.pytorch.org/docs/main/generated/torch.ones.html
                                                        #      https://docs.pytorch.org/docs/main/generated/torch.set_default_dtype.html#torch.set_default_dtype
b = torch.ones((2, 3), dtype=torch.int16)
c = torch.rand((2, 3), dtype=torch.float64) * 20.
d = b.to(torch.int32)       # b를 int32로 변환.

print(a.dtype)
print(b)
print(c)
print(d)
```

```python
double_d = torch.ones(10, 2, dtype=torch.double)     # float64
short_e = torch.tensor([[1, 2]], dtype=torch.short)  # int16

double_d = torch.zeros(10, 2).double()      # float32 -> float64
short_e = torch.ones(10, 2).short()         # float32 -> int16

# 상동. 다른 방법.
double_d = torch.zeros(10, 2).to(torch.double)
short_e = torch.ones(10, 2).to(dtype=torch.short)

# 상동. 다른 방법.
double_d = torch.zeros(10, 2).type(torch.double)
short_e = torch.ones(10, 2).type(dtype=torch.short)

print(double_d.dtype)
print(short_e.dtype)

# 참고: 기본 자료형은 float32.
double_d = torch.zeros(10, 2)
short_e = torch.ones(10, 2)

print(double_d.dtype)
print(short_e.dtype)
```

```python
# 연산 시 자동 형변환.
# float64와 int16 중 더 넓은 도메인을 갖는 쪽으로 변환.
# 따라서, float64 * int16 -> float64
#
# 파이토치는 C++ 기반이므로 C++의 자동 형변환 규칙을 따를 것으로 예상됨.
# 참고: intx, inty -> intx (x > y)
#       floatx, floaty -> floatx (x > y)
#       signed, unsigned -> unsigned
#       integer, float -> float
double_f = torch.rand(5, dtype=torch.double)
short_g = double_f.to(torch.short)
print((double_f * short_g).dtype)
```

# f_tensor_operations

```python
# 사칙 연산은 member-wise로 수행됨.
# e.g. [1, 2] + [3, 4] = [4, 6]
t1 = torch.ones(size=(2, 3))
t2 = torch.ones(size=(2, 3))
t3 = torch.add(t1, t2)
t4 = t1 + t2        # 참고: 연산자 재정의로 add를 사용하지 않고 연산자를 사용할 수 있다.

print(t3)
print(t4)
```

```python
# 뺄셈, 곱셈, 나숫셈도 상동.
t5 = torch.sub(t1, t2)
t6 = t1 - t2
print(t5)
print(t6)

print("#" * 30)

t7 = torch.mul(t1, t2)
t8 = t1 * t2
print(t7)
print(t8)

print("#" * 30)

t9 = torch.div(t1, t2)
t10 = t1 / t2
print(t9)
print(t10)
```

```python
# member-wise이므로 두 텐서의 크기가 다르면 오류.
try:
    x1 = torch.ones(size=(2, 4))
    x2 = torch.ones(size=(2, 3))
    x3 = x1 + x2
except:
    print('tensor size does not match')
```

```python
# member-wise이므로 아래와 같이 내적을 구하는 데 사용할 수 있다.
v1 = torch.tensor([1, 2])
v2 = torch.tensor([3, 4])

print(sum(v1.mul(v2)))
print(sum(v1 * v2))
print(v1.dot(v2))     # 물론 파이토치에는 내적을 구하는 함수가 별도로 존재한다.
```

# g_tensor_operations_mm

```python
# 내적 연산.
t1 = torch.dot(
  torch.tensor([2, 3]), torch.tensor([2, 1])
)

# 내적의 결과는 스칼라이므로 크기는 []가 됨.
print(t1, t1.size())
```

```python
# 텐서의 행렬곱 연산.
# 행렬곱이므로 텐서의 크기는 (n, m) * (m, k)여야 함.
t2 = torch.randn(2, 3)
t3 = torch.randn(3, 2)
t4 = torch.mm(t2, t3)
print(t4, t4.size())
```

```python
# 행렬곱 연산이지만 batch 처리.
t5 = torch.randn(10, 3, 4)
t6 = torch.randn(10, 4, 5)
t7 = torch.bmm(t5, t6)

print(t7.size())

# 아래 두 코드는 동등한 효과를 갖는다.
torch_batched = torch.bmm(t5, t6)
manually_batched = torch.stack([torch.mm(x, y) for x, y in zip(t5, t6)])

# 부동소수점 자료형이므로 eps를 도입하여 값 비교.
print(torch.all(torch_batched - manually_batched) < 1e-5)
```

# h_tensor_operations_matmut

```python
# ref: https://docs.pytorch.org/docs/stable/generated/torch.matmul.html

# vector x vector: dot product
t1 = torch.randn(3)
t2 = torch.randn(3)
print(torch.matmul(t1, t2).size())  # torch.Size([])
```

```python
# matrix x vector: broadcasted dot
t3 = torch.randn(3, 4)
t4 = torch.randn(4)
print(torch.matmul(t3, t4).size())  # torch.Size([3])

# 행렬 * 벡터 -> 행렬 * 열벡터로 간주함.
```

```python
# vector x matrix: broadcasted dot
x = torch.randn(3)
y = torch.randn(3, 4)
print(torch.matmul(x, y).size())  # torch.Size([3])

# 벡터 * 행렬 -> 행벡터 * 행렬로 간주함.
```

```python
# 이외의 3차원 이상의 텐서가 주어지면 batch 처리함.

# batched matrix x vector: broadcasted dot
t5 = torch.randn(10, 3, 4)
t6 = torch.randn(4)

# batched matrix x batched matrix: bmm
t7 = torch.randn(10, 3, 4)
t8 = torch.randn(10, 4, 5)
print(torch.matmul(t7, t8).size())  # torch.Size([10, 3, 5])

# batched matrix x matrix: bmm
t9 = torch.randn(10, 3, 4)
t10 = torch.randn(4, 5)
print(torch.matmul(t9, t10).size())  # torch.Size([10, 3, 5])
```

# i_tensor_broadcasting

```python
# broadcasting이란?
#     서로 다른 크기를 갖는 두 텐서의 연산을 가능케 하기 위해 텐서의 크기를 변형하는 것.
#     ref: https://docs.pytorch.org/docs/stable/notes/broadcasting.html
#
#     Each tensor has at least one dimension.
#     When iterating over the dimension sizes, starting at the trailing dimension, the dimension sizes must either be equal, one of them is 1, or one of them does not exist.

t1 = torch.tensor([1.0, 2.0, 3.0])
t2 = 2.0

# [1, 2, 3] * 2
# -> [1, 2, 3] * [2, 2, 2]
# -> [2, 4, 6]
print(t1 * t2)
print((t1 * t2).size())
```

```python
t3 = torch.tensor([[0, 1], [2, 4], [10, 10]])
t4 = torch.tensor([4, 5])

# [
#     [0, 1],
#     [2, 4],    -  [4, 5]
#     [10, 10],
# ]
# ->
# [                 [
#     [0, 1],           [4, 5],
#     [2, 4],    -      [4, 5],
#     [10, 10],         [4, 5],
# ]                 ]
# ->
# [
#     [-4, -5],
#     [-2, -1],
#     [6, 5],
# ]
print(t3 - t4)
print((t3 - t4).size())
```

```python
# 상동.
t5 = torch.tensor([[1., 2.], [3., 4.]])
print(t5 + 2.0)  # t5.add(2.0)
print(t5 - 2.0)  # t5.sub(2.0)
print(t5 * 2.0)  # t5.mul(2.0)
print(t5 / 2.0)  # t5.div(2.0)
```

```python
# 스칼라로 나누므로 x가 텐서여도 broadcasting되어 연산 가능.
def normalize(x):
    return x / 255

t6 = torch.randn(3, 28, 28)
print(normalize(t6).size())
```

```python
t7 = torch.tensor([[1, 2], [0, 3]])  # torch.Size([2, 2])
t8 = torch.tensor([[3, 1]])  # torch.Size([1, 2])
t9 = torch.tensor([[5], [2]])  # torch.Size([2, 1])
t10 = torch.tensor([7])  # torch.Size([1])

# (2, 2) + (1, 2) -> (2, 2)
print(t7 + t8)   # >>> tensor([[4, 3], [3, 4]])

# (2, 2) + (2, 1) -> (2, 2)
print(t7 + t9)   # >>> tensor([[6, 7], [2, 5]])

# (1, 2) + (2, 1) -> (2, 2)
print(t8 + t9)   # >>> tensor([[8, 6], [5, 3]])

# (2, 2) + (1) -> (2, 2)
print(t7 + t10)  # >>> tensor([[ 8, 9], [ 7, 10]])
```

```python
# (4, 3, 2) * (3, 2) -> (4, 3, 2)
t11 = torch.ones(4, 3, 2)
t12 = t11 * torch.rand(3, 2)  # 3rd & 2nd dims identical to t11, dim 0 absent
print(t12.shape)

# (4, 3, 2) * (3, 1) -> (4, 3, 2)
t13 = torch.ones(4, 3, 2)
t14 = t13 * torch.rand(3, 1)  # 3rd dim = 1, 2nd dim is identical to t13
print(t14.shape)

# (4, 3, 2) * (1, 2) -> (4, 3, 2)
t15 = torch.ones(4, 3, 2)
t16 = t15 * torch.rand(1, 2)  # 3rd dim is identical to t15, 2nd dim is 1
print(t16.shape)

# (5, 3, 4, 1) + (3, 1, 1) -> (5, 3, 4, 1)
t17 = torch.ones(5, 3, 4, 1)
t18 = torch.rand(3, 1, 1)  # 2nd dim is identical to t17, 3rd and 4th dims are 1
print((t17 + t18).size())
```

```python
# (5, 1, 4, 1) + (3, 1, 1) -> (5, 3, 4, 1)
t19 = torch.empty(5, 1, 4, 1)
t20 = torch.empty(3, 1, 1)
print((t19 + t20).size())  # torch.Size([5, 3, 4, 1])

# (1) + (3, 1, 7) -> (3, 1, 7)
t21 = torch.empty(1)
t22 = torch.empty(3, 1, 7)
print((t21 + t22).size())  # torch.Size([3, 1, 7])

# (3, 3, 3) + (3, 1, 3) -> (3, 3, 3)
t23 = torch.ones(3, 3, 3)
t24 = torch.ones(3, 1, 3)
print((t23 + t24).size())  # torch.Size([3, 3, 3])

try:
    t25 = torch.empty(5, 2, 4, 1)
    t26 = torch.empty(3, 1, 1)
    print((t25 + t26).size())
except:
    print('(5, <2>, 4, 1) + (<3>, 1, 1) -> (5, <x>, 4, 1)')
    print('2 and 1 is not equal in dim 1')
```

```python
# (4) * sclar -> (4)
t27 = torch.ones(4) * 5
print(t27)  # >>> tensor([ 5, 5, 5, 5])

# (4) ^ scalar -> (4)
t28 = torch.pow(t27, 2)
print(t28)  # >>> tensor([ 25, 25, 25, 25])

# (4) ^ (4) -> (4)
exp = torch.arange(1., 5.)  # tensor([ 1.,  2.,  3.,  4.])
a = torch.arange(1., 5.)  # tensor([ 1.,  2.,  3.,  4.])
t29 = torch.pow(a, exp)
print(t29)  # >>> tensor([   1.,    4.,   27.,  256.])
```
