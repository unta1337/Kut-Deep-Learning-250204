# b_tensor_initialization_copy

```python
l1 = [1, 2, 3]
t1 = torch.Tensor(l1)

l2 = [1, 2, 3]
t2 = torch.tensor(l2)

l3 = [1, 2, 3]
t3 = torch.as_tensor(l3)

# 각각의 텐서는 원본 배열을 복사해 생성됨.
# 즉, 원본을 수정해도 복사된 텐서는 변하지 않음.
l1[0] = 100
l2[0] = 100
l3[0] = 100

print(t1)
print(t2)
print(t3)
```

```python
l4 = np.array([1, 2, 3])
t4 = torch.Tensor(l4)

l5 = np.array([1, 2, 3])
t5 = torch.tensor(l5)

l6 = np.array([1, 2, 3])
t6 = torch.as_tensor(l6)

# 넘파이 배열도 기본적으로 복사 후 텐서를 생성함.
# 단, as_tensor로 불러오면 참조로 생성함.
# 즉, 원본 넘파이 배열을 수정하면 텐서도 변경됨.
l4[0] = 100
l5[0] = 100
l6[0] = 100

print(t4)
print(t5)
print(t6)
```

# c_tensor_initialization_constant_values

```python
# size 크기의 텐서를 1로 초기화.
t1 = torch.ones(size=(5,))  # or torch.ones(5)

# input과 동일한 크기의 텐서를 1로 초기화.
t1_like = torch.ones_like(input=t1)

print(t1)  # >>> tensor([1., 1., 1., 1., 1.])
print(t1_like)  # >>> tensor([1., 1., 1., 1., 1.])
```

```python
# 상동. 단, 1 대신 0으로 초기화.
t2 = torch.zeros(size=(6,))  # or torch.zeros(6)
t2_like = torch.zeros_like(input=t2)
print(t2)  # >>> tensor([0., 0., 0., 0., 0., 0.])
print(t2_like)  # >>> tensor([0., 0., 0., 0., 0., 0.])
```

```python
# 상동. 단, 초기화되지 않은 쓰레기값이 들어갈 수 있음.
# ref: https://docs.pytorch.org/docs/stable/generated/torch.empty.html
t3 = torch.empty(size=(4,))  # or torch.zeros(4)
t3_like = torch.empty_like(input=t3)
print(t3)  # >>> tensor([0., 0., 0., 0.])
print(t3_like)  # >>> tensor([0., 0., 0., 0.])
```

# e_tensor_type_conversion

```python
a = torch.ones((2, 3))                                  # 기본 자료형: float32
                                                        # 기본 자료형은 전역 기본 자료형을 사용한다.
                                                        # 이때 전역 기본 자료형은 최초 설치 시 float32로 설정된다.
                                                        # ref: https://docs.pytorch.org/docs/main/generated/torch.ones.html
                                                        #      https://docs.pytorch.org/docs/main/generated/torch.set_default_dtype.html#torch.set_default_dtype
b = torch.ones((2, 3), dtype=torch.int16)
c = torch.rand((2, 3), dtype=torch.float64) * 20.
d = b.to(torch.int32)       # b를 int32로 변환.

print(a.dtype)
print(b)
print(c)
print(d)
```

```python
double_d = torch.ones(10, 2, dtype=torch.double)     # float64
short_e = torch.tensor([[1, 2]], dtype=torch.short)  # int16

double_d = torch.zeros(10, 2).double()      # float32 -> float64
short_e = torch.ones(10, 2).short()         # float32 -> int16

# 상동. 다른 방법.
double_d = torch.zeros(10, 2).to(torch.double)
short_e = torch.ones(10, 2).to(dtype=torch.short)

# 상동. 다른 방법.
double_d = torch.zeros(10, 2).type(torch.double)
short_e = torch.ones(10, 2).type(dtype=torch.short)

print(double_d.dtype)
print(short_e.dtype)

# 참고: 기본 자료형은 float32.
double_d = torch.zeros(10, 2)
short_e = torch.ones(10, 2)

print(double_d.dtype)
print(short_e.dtype)
```

```python
# 연산 시 자동 형변환.
# float64와 int16 중 더 넓은 도메인을 갖는 쪽으로 변환.
# 따라서, float64 * int16 -> float64
#
# 파이토치는 C++ 기반이므로 C++의 자동 형변환 규칙을 따를 것으로 예상됨.
# 참고: intx, inty -> intx (x > y)
#       floatx, floaty -> floatx (x > y)
#       signed, unsigned -> unsigned
#       integer, float -> float
double_f = torch.rand(5, dtype=torch.double)
short_g = double_f.to(torch.short)
print((double_f * short_g).dtype)
```

# f_tensor_operations

```python
# 사칙 연산은 member-wise로 수행됨.
# e.g. [1, 2] + [3, 4] = [4, 6]
t1 = torch.ones(size=(2, 3))
t2 = torch.ones(size=(2, 3))
t3 = torch.add(t1, t2)
t4 = t1 + t2        # 참고: 연산자 재정의로 add를 사용하지 않고 연산자를 사용할 수 있다.

print(t3)
print(t4)
```

```python
# 뺄셈, 곱셈, 나숫셈도 상동.
t5 = torch.sub(t1, t2)
t6 = t1 - t2
print(t5)
print(t6)

print("#" * 30)

t7 = torch.mul(t1, t2)
t8 = t1 * t2
print(t7)
print(t8)

print("#" * 30)

t9 = torch.div(t1, t2)
t10 = t1 / t2
print(t9)
print(t10)
```

```python
# member-wise이므로 두 텐서의 크기가 다르면 오류.
try:
    x1 = torch.ones(size=(2, 4))
    x2 = torch.ones(size=(2, 3))
    x3 = x1 + x2
except:
    print('tensor size does not match')
```

```python
# member-wise이므로 아래와 같이 내적을 구하는 데 사용할 수 있다.
v1 = torch.tensor([1, 2])
v2 = torch.tensor([3, 4])

print(sum(v1.mul(v2)))
print(sum(v1 * v2))
print(v1.dot(v2))     # 물론 파이토치에는 내적을 구하는 함수가 별도로 존재한다.
```

# g_tensor_operations_mm

```python
# 내적 연산.
t1 = torch.dot(
  torch.tensor([2, 3]), torch.tensor([2, 1])
)

# 내적의 결과는 스칼라이므로 크기는 []가 됨.
print(t1, t1.size())
```

```python
# 텐서의 행렬곱 연산.
# 행렬곱이므로 텐서의 크기는 (n, m) * (m, k)여야 함.
t2 = torch.randn(2, 3)
t3 = torch.randn(3, 2)
t4 = torch.mm(t2, t3)
print(t4, t4.size())
```

```python
# 행렬곱 연산이지만 batch 처리.
t5 = torch.randn(10, 3, 4)
t6 = torch.randn(10, 4, 5)
t7 = torch.bmm(t5, t6)

print(t7.size())

# 아래 두 코드는 동등한 효과를 갖는다.
torch_batched = torch.bmm(t5, t6)
manually_batched = torch.stack([torch.mm(x, y) for x, y in zip(t5, t6)])

# 부동소수점 자료형이므로 eps를 도입하여 값 비교.
print(torch.all(torch_batched - manually_batched) < 1e-5)
```

# h_tensor_operations_matmut

```python
# ref: https://docs.pytorch.org/docs/stable/generated/torch.matmul.html

# vector x vector: dot product
t1 = torch.randn(3)
t2 = torch.randn(3)
print(torch.matmul(t1, t2).size())  # torch.Size([])
```

```python
# matrix x vector: broadcasted dot
t3 = torch.randn(3, 4)
t4 = torch.randn(4)
print(torch.matmul(t3, t4).size())  # torch.Size([3])

# 행렬 * 벡터 -> 행렬 * 열벡터로 간주함.
```

```python
# vector x matrix: broadcasted dot
x = torch.randn(3)
y = torch.randn(3, 4)
print(torch.matmul(x, y).size())  # torch.Size([3])

# 벡터 * 행렬 -> 행벡터 * 행렬로 간주함.
```

```python
# 이외의 3차원 이상의 텐서가 주어지면 batch 처리함.

# batched matrix x vector: broadcasted dot
t5 = torch.randn(10, 3, 4)
t6 = torch.randn(4)

# batched matrix x batched matrix: bmm
t7 = torch.randn(10, 3, 4)
t8 = torch.randn(10, 4, 5)
print(torch.matmul(t7, t8).size())  # torch.Size([10, 3, 5])

# batched matrix x matrix: bmm
t9 = torch.randn(10, 3, 4)
t10 = torch.randn(4, 5)
print(torch.matmul(t9, t10).size())  # torch.Size([10, 3, 5])
```
