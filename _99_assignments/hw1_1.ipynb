# a_tensor_initialization

```python
# 자료형 기본값: float32
t1 = torch.Tensor([1, 2, 3], device='cpu')      # cpu 위에서 연산함에 유의.
print(t1.dtype)   # >>> torch.float32
print(t1.device)  # >>> cpu
print(t1.requires_grad)  # >>> False
print(t1.size())  # torch.Size([3])
print(t1.shape)   # torch.Size([3])

# 아래와 같이 cpu/gpu 간 상호 변환 가능.
t1_cpu = t1.cpu()
t1_cuda = t1.cuda()
```

```python
# 자료형 기본값: float32
t2 = torch.tensor([1, 2, 3], device='cpu')
print(t2.dtype)  # >>> torch.int64
print(t2.device)  # >>> cpu
print(t2.requires_grad)  # >>> False
print(t2.size())  # torch.Size([3])
print(t2.shape)  # torch.Size([3])
```

# b_tensor_initialization_copy

```python
l1 = [1, 2, 3]
t1 = torch.Tensor(l1)

l2 = [1, 2, 3]
t2 = torch.tensor(l2)

l3 = [1, 2, 3]
t3 = torch.as_tensor(l3)

# 각각의 텐서는 원본 배열을 복사해 생성됨.
# 즉, 원본을 수정해도 복사된 텐서는 변하지 않음.
l1[0] = 100
l2[0] = 100
l3[0] = 100

print(t1)
print(t2)
print(t3)
```

```python
l4 = np.array([1, 2, 3])
t4 = torch.Tensor(l4)

l5 = np.array([1, 2, 3])
t5 = torch.tensor(l5)

l6 = np.array([1, 2, 3])
t6 = torch.as_tensor(l6)

# 넘파이 배열도 기본적으로 복사 후 텐서를 생성함.
# 단, as_tensor로 불러오면 참조로 생성함.
# 즉, 원본 넘파이 배열을 수정하면 텐서도 변경됨.
l4[0] = 100
l5[0] = 100
l6[0] = 100

print(t4)
print(t5)
print(t6)
```

# c_tensor_initialization_constant_values

```python
# size 크기의 텐서를 1로 초기화.
t1 = torch.ones(size=(5,))  # or torch.ones(5)

# input과 동일한 크기의 텐서를 1로 초기화.
t1_like = torch.ones_like(input=t1)

print(t1)  # >>> tensor([1., 1., 1., 1., 1.])
print(t1_like)  # >>> tensor([1., 1., 1., 1., 1.])
```

```python
# 상동. 단, 1 대신 0으로 초기화.
t2 = torch.zeros(size=(6,))  # or torch.zeros(6)
t2_like = torch.zeros_like(input=t2)
print(t2)  # >>> tensor([0., 0., 0., 0., 0., 0.])
print(t2_like)  # >>> tensor([0., 0., 0., 0., 0., 0.])
```

```python
# 상동. 단, 초기화되지 않은 쓰레기값이 들어갈 수 있음.
# ref: https://docs.pytorch.org/docs/stable/generated/torch.empty.html
t3 = torch.empty(size=(4,))  # or torch.zeros(4)
t3_like = torch.empty_like(input=t3)
print(t3)  # >>> tensor([0., 0., 0., 0.])
print(t3_like)  # >>> tensor([0., 0., 0., 0.])
```
